```python
from scipy.optimize import linear_sum_assignment
import numpy as np

# Matriks biaya pengiriman dari setiap pabrik ke setiap toko
costs = [
    [1, 9, 13, 36, 51],  # Biaya dari Pabrik 1
    [24, 12, 16, 20, 1],  # Biaya dari Pabrik 2
    [14, 33, 1, 23, 26]   # Biaya dari Pabrik 3
]
capacity = [50, 100, 150]  # Kapasitas Pabrik 1, Pabrik 2, Pabrik 3
requirement = [100, 70, 50, 40, 40]  # Permintaan Toko 1, Toko 2, Toko 3, Toko 4, Toko 5

# Inisialisasi matriks alokasi dengan nol
allocation = np.zeros((len(capacity), len(requirement)))
# Implementasi Metode Least Cost
while True:
    min_cost = float('inf')
    min_i = -1
    min_j = -1
    # Mencari biaya minimum
    for i in range(len(capacity)):
        for j in range(len(requirement)):
            if capacity[i] > 0 and requirement[j] > 0 and costs[i][j] < min_cost:
                min_cost = costs[i][j]
                min_i = i
                min_j = j
    if min_i == -1 or min_j == -1:
        break
    # Menentukan jumlah pengiriman
    send_amount = min(capacity[min_i], requirement[min_j])
    allocation[min_i][min_j] = send_amount
    # Mengurangi sisa kapasitas pabrik dan permintaan toko
    capacity[min_i] -= send_amount
    requirement[min_j] -= send_amount
# Mencetak hasil Least Cost
print("Alokasi setelah metode Least Cost:")
print(allocation)
total_cost_least_cost = np.sum(allocation * np.array(costs))
print(f"\nTotal Biaya setelah metode Least Cost: {total_cost_least_cost}")
```

```python
import numpy as np
from scipy.optimize import linprog

# Matriks biaya pengiriman dari setiap pabrik ke setiap toko
costs = np.array([[1, 9, 13, 36, 51], # Biaya dari Pabrik 1
                  [24, 12, 16, 20, 1], # Biaya dari Pabrik 2
                  [14, 33, 1, 23, 26]]) # Biaya dari Pabrik 3
capacity = np.array([50, 100, 150]) # Kapasitas Pabrik 1, Pabrik 2, Pabrik 3
requirement = np.array([100, 70, 50, 40, 40]) # Permintaan Toko 1, Toko 2, Toko 3, Toko 4, Toko 5

# Pembentukan Matriks dan Array Kendala
c = costs.flatten()
A_eq = np.zeros((len(capacity) + len(requirement), len(c)))
b_eq = np.zeros(len(capacity) + len(requirement))
# Pengisian Matriks Kendala
for i in range(len(capacity)):
    A_eq[i, i*len(requirement):(i+1)*len(requirement)] = 1
    b_eq[i] = capacity[i]
for j in range(len(requirement)):
    A_eq[len(capacity)+j, j::len(requirement)] = 1
    b_eq[len(capacity)+j] = requirement[j]
# Pemecahan Masalah dengan Linear Programming
res = linprog(c, A_eq=A_eq, b_eq=b_eq, method='highs')
# Mencetak Solusi Optimal
print('Shipping cost:')
print(res.x.reshape(costs.shape))
print('\nSolusi Optimal:')
print(res.fun)
```
